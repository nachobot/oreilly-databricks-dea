## Ingestion Methods
- Batch
  All data is re-ingested each time the pipeline runs
  - CTAS
  - spark.read.load()

Incremental Batch
- Only new data is ingested 
- Faster and more resource efficient ingestion

1. COPY INTO
2. spark.readStream (**auto Loader** with timed trigger)
3. Declarative Pipeline (CREATE OR **REFRESH** STREAMING TABLE)


## Streaming
- Continuously load data rows or batches of data rows at it is generated
    Query as it arrives in near real-time
- Micro-batch 
    process small batches - very short, frequent intervals
    - spark.readStream (Auto Loader with continuous trigger)
    - Declarative Pipelines (trigger mode continuous)

  Enables
  - low-latency analysis
  - Immediate action
