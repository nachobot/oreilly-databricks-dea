# 9.1 Liquid Clustering

- Replaces - Traditional Table partitioning and ZORDER
- Allows for - Redefinition of cluster keys without rewriting existing data
- Applies to Streaming Tables and Materialized Views

```python
%sql
CREATE TABLE table1 (
  col0 INT,
  col1 STRING
) USING DELTA
CLUSTER BY (col0);
```

Auto liquid clustering

```python
%sql
CREATE TABLE table2 (
  col0 INT,
  col1 STRING
) USING DELTA
CLUSTER BY AUTO;
```

```python
%sql
ALTER TABLE table1 CLUSTER BY AUTO
```

```python
%sql
DESCRIBE TABLE table1
```
